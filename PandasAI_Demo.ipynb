## 🔑 Step 2: API Key Setup

**Gemini API Key পেতে:**

2. Google account দিয়ে login করুন

4. Key copy করে নিচের cell-এ paste করুন

⚠️ **Important:** API key কাউকে দেবেন না, public repo-তে commit করবেন না!

```python
import pandas as pd
import pandasai as pai
from pandasai_litellm.litellm import LiteLLM

# 🔥 এখানে আপনার Gemini API key paste করুন
GEMINI_API_KEY = "YOUR_GEMINI_API_KEY_HERE"  # Replace with your key

# Configure LLM (Large Language Model)
llm = LiteLLM(
    model="gemini/gemini-2.0-flash",
    api_key=GEMINI_API_KEY
)

# Set global config for PandasAI
pai.config.set({
    "llm": llm,
    "enable_cache": True,        # Cache results for faster responses
    "verbose": True,             # Show generated code
    "save_charts": True,         # Save visualizations
    "save_charts_path": "./charts"  # Charts folder
})

print("✅ PandasAI configured successfully with Gemini!")
print("এখন আপনি plain English-এ data নিয়ে chat করতে পারবেন 🚀")
```

## 📊 Step 3: Load Your Data

আপনার marketing campaign data load করি। আমি sample data তৈরি করছি — আপনি আপনার actual CSV file দিয়ে replace করবেন।

```python
# Option 1: Load from CSV file (আপনার actual file path দিন)
# df = pd.read_csv("path/to/your/marketing_campaign_data.csv")

# Option 2: Sample data for demo (যদি file না থাকে)
import numpy as np
np.random.seed(42)

sample_data = {
    'Campaign_ID': [f'CMP-{i:05d}' for i in range(1, 101)],
    'Channel': np.random.choice(['Facebook', 'Google Ads', 'TikTok', 'Email'], 100),
    'Impressions': np.random.randint(1000, 100000, 100),
    'Clicks': np.random.randint(50, 5000, 100),
    'Spend': np.random.uniform(100, 10000, 100).round(2),
    'Conversions': np.random.randint(1, 500, 100),
    'CTR': np.random.uniform(0.5, 5.0, 100).round(2),
    'CPC': np.random.uniform(0.5, 3.0, 100).round(2),
    'Conversion_Rate': np.random.uniform(1, 15, 100).round(2),
    'Cost_Per_Conversion': np.random.uniform(5, 50, 100).round(2)
}

df = pd.DataFrame(sample_data)

# Convert to PandasAI SmartDataFrame
smart_df = pai.DataFrame(df)

print(f"✅ Data loaded: {len(df)} rows, {len(df.columns)} columns")
print("\n📋 Data Preview:")
print(df.head())
print("\n🎉 Ready to chat with your data!")
```

## 💬 Step 4: Basic Chat Examples

```python
# Basic information about the dataset
response = smart_df.chat("How many rows and columns are there in this dataset?")
print("📊 Dataset Info:", response)
```

```python
# Get column names and types
response = smart_df.chat("What are the column names and their data types?")
print("📝 Columns Info:")
print(response)
```

## 💡 Tips & Tricks

### ✅ Good Query Examples:
- "Show me campaigns with spend above $5000"
- "Create a bar chart of conversion rate by channel"
- "Calculate the correlation between CPC and conversion rate"
- "Find outliers in the Clicks column using IQR method"
- "Group by Channel and show average metrics"

### ❌ Avoid:
- Very vague questions ("analyze the data")
- Questions about data you don't have
- Asking for personal information

### 🔧 Useful Commands:
- `smart_df.last_code_executed` — See generated code
- `pai.config.set({"verbose": True})` — Show code execution
- `pai.config.set({"save_charts": True})` — Auto-save charts
